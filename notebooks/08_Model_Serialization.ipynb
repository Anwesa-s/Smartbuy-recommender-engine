# Model Serialization

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import pickle
```

```python
recommendation_df = pd.read_csv(
    "../data/recommendation_dataset.csv"
)
```

```python
recommendation_df['combined_features'] = (
    recommendation_df['category']
    + " "
    +
    recommendation_df['about_product']
)
```

```python
tfidf = TfidfVectorizer(
    stop_words='english'
)

tfidf_matrix = tfidf.fit_transform(
    recommendation_df['combined_features']
)
```

```python
cosine_sim = cosine_similarity(
    tfidf_matrix,
    tfidf_matrix
)
```

```python
indices = pd.Series(
    recommendation_df.index,
    index=recommendation_df['product_name']
)
```

```python
with open(
    "../models/tfidf.pkl",
    "wb"
) as f:
    pickle.dump(
        tfidf,
        f
    )
```

```python
with open(
    "../models/cosine_sim.pkl",
    "wb"
) as f:
    pickle.dump(
        cosine_sim,
        f
    )
```

```python
with open(
    "../models/indices.pkl",
    "wb"
) as f:
    pickle.dump(
        indices,
        f
    )
```
